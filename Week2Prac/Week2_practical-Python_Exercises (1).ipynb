### Task 2. A few Simple Python Exercises 

1. Given the following dictionary:  
`student_grades = {'Adam':'CR', 'Leo':'HD', 'Max':'D', 'Nicole':'CR', 'Oscar':'P'}`  
print all key-value pairs:  
    - `('Adam', 'CR')` 
    - `('Leo', 'HD')`
    - `('Max', 'D')`
    - `('Nicole', 'CR')` 
    - `('Oscar', 'P')` 

2. Given the same dictionary as above, print all keys.

3. Given the same dictionary as above, print all values. 

4. Write a Python program that prints the current date and time in the same format as the following example: `2021-07-20 20:38:48`. Note that you have to import the `datetime` module and use the instance method: `date.strftime(format)` for this purpose. Check the Python documentation for details. 

5. Write a Python function `max_number_in_list(list)` that returns the largest number in a list. 

6.  Given a list of seven fruits: `['Apple', 'Banana', 'Fig', 'Pear', 'Plum', 'Peach', 'Raspberry']`, write Python code that always removes the first fruit, the fifth fruit, and the sixth fruit from such a list and returns the following list: `['Banana', 'Fig', 'Pear', 'Raspberry']`. Note that for the following list: `['Banana', 'Apple', 'Pear', 'Plum', 'Peach', 'Fig', 'Raspberry']`, the result should look as follows: `['Apple', 'Pear', 'Plum', 'Raspberry']`. 

7. Write a Python function that takes a list of words as input and displays the longest word. 

8. Given the following two sets `{green, blue}` and `{blue, yellow}`, write Python code that calculates (a) the intersection and (b) the union of these two sets. 

If you have problems with these tasks, then please ask your practical tutor for help.
